```python
import torch
x = torch.tensor(2.0)
y = torch.tensor(3.0, requires_grad=True)

y_true = torch.tensor(10.0)
pred = x * y
loss = (pred - y_true)**2

print(loss)
loss.backward()
print(y.grad) 
```

```python
import torch
# input fixed 
x = torch.tensor(2.0)

#lernable number
w = torch.tensor(3.0, requires_grad=True)

#correct answer
y_true = torch.tensor(10.0)

#learning rate
lr = 0.1

y_pred = w * x
loss = (y_pred - y_true) ** 2

loss.backward()

print("Before update:", w.item())
print("loss:", loss.item())
print("Gradient:", w.grad.item())

with torch.no_grad(): # pytorch doesnt track this update 
    w-= lr*w.grad #update the weight
    
w.grad.zero_() 

print("After update:", w.item())

```
